#### График Кюри (пункты 4 и 5 , остальные пункты в фото)
$$K(E)=\sqrt{\frac{1}{C(E)}\frac{d\,\Gamma}{dE}} = \sqrt{E_{\nu}\sqrt{E_{\nu}^2 - m^2_{\nu}}}$$

```python
import numpy as np
import matplotlib.pyplot as plt

E_0 = 18.575e3  # эВ

def KurieFunc(E, m_nu):
    """Kurie Function"""
    E_nu = E_0 - E
    K = np.zeros_like(E)
    K[E_nu >= m_nu] = np.sqrt(E_nu[E_nu >= m_nu] * np.sqrt(E_nu[E_nu >= m_nu]**2 - m_nu**2))
    return K


E = np.linspace(E_0 - 5, E_0, 10000)
K_0 = KurieFunc(E, 0.)
K_1 = KurieFunc(E, 1.)

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

ax = axs[0]
ax.plot(E, K_0, label=r'$m_\nu = 0$')
ax.plot(E, K_1, label=r'$m_\nu = 1$ eV')

ax.axvline(E_0, color='red', ls='--', label=r'$E = E_0$ (конец для $m_\nu=0$)')
ax.axvline(E_0 - 1., color='green', ls='--', label=r'$E = E_0 - 1$ eV (конец для $m_\nu=1$ eV)')

ax.set_xlabel('E, eV')
ax.set_ylabel('K(E), eV')
ax.set_title('Линейный')
ax.legend()
ax.grid(alpha=0.6)

ax = axs[1]
ax.plot(E, K_0, label=r'$m_\nu = 0$')
ax.plot(E, K_1, label=r'$m_\nu = 1$ eV')

ax.axvline(E_0, color='red', ls='--', label=r'$E = E_0$ (конец для $m_\nu=0$)')
ax.axvline(E_0 - 1., color='green', ls='--', label=r'$E = E_0 - 1$ eV (конец для $m_\nu=1$ eV)')

ax.set_yscale("log")
ax.set_xlabel('E, eV')
ax.set_title('Логарифмический')
ax.legend()
ax.grid(alpha=0.6)

fig.suptitle('Функция Кюри последних 5 эВ спектра')
plt.tight_layout()
plt.show()
```

#### Пункт 5. Каждое массовое состояние даёт излом спектра при энергии равной массе: $E_i = E_0 - m_i$. Расстояние между изломами двух состояний равно разнице масс этих состояний: $\Delta E_{ij} = \Delta m_{ij}$.
