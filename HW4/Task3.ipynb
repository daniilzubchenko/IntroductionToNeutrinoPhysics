```python
import numpy as np

m = 511e3 # eV
alpha = 1/137

def F(Z, K):
    K = np.asarray(K)
    E = K + m
    p = np.sqrt(E**2 - m**2)
    eta = Z * alpha * E / p
    return (2*np.pi * eta) / (1 - np.exp(-2*np.pi*eta))

K = [10e3, 100e3, 1e6] # eV
Z = [29, -29]
for z in Z:
    print("Z", z)
    print(F(z, K))
```
